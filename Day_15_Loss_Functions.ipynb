```python
import tensorflow as tf

actual = tf.constant([1.0, 0.0, 1.0])
prediction = tf.constant([0.9, 0.1, 0.8])

mse = tf.keras.losses.MeanSquaredError()
mae = tf.keras.losses.MeanAbsoluteError()
huber = tf.keras.losses.Huber()

print("MSE:", mse(actual, prediction).numpy())
print("MAE:", mae(actual, prediction).numpy())
print("Huber:", huber(actual, prediction).numpy())
```

```python
bce = tf.keras.losses.BinaryCrossentropy()

print("BCE:", bce(actual, prediction).numpy())
```

```python
import tensorflow as tf

print("========== DAY 15: LOSS FUNCTIONS ==========\n")

# ============================================================
# 1. MSE, MAE, HUBER
# ============================================================

actual_reg = tf.constant([1.0, 0.0, 1.0])
prediction_reg = tf.constant([0.9, 0.1, 0.8])

mse = tf.keras.losses.MeanSquaredError()
mae = tf.keras.losses.MeanAbsoluteError()
huber = tf.keras.losses.Huber()

mse_result = mse(actual_reg, prediction_reg)
mae_result = mae(actual_reg, prediction_reg)
huber_result = huber(actual_reg, prediction_reg)

print("---- Regression Losses ----")
print("MSE   :", mse_result.numpy())
print("MAE   :", mae_result.numpy())
print("Huber :", huber_result.numpy())

# Manual calculations
errors = actual_reg - prediction_reg

manual_mse = tf.reduce_mean(tf.square(errors))
manual_mae = tf.reduce_mean(tf.abs(errors))

print("\nManual MSE :", manual_mse.numpy())
print("Manual MAE :", manual_mae.numpy())


# ============================================================
# 2. BINARY CROSS-ENTROPY
# ============================================================

actual_binary = tf.constant([1.0, 0.0, 1.0])
prediction_binary = tf.constant([0.9, 0.1, 0.8])

bce = tf.keras.losses.BinaryCrossentropy()

bce_result = bce(actual_binary, prediction_binary)

print("\n---- Binary Cross-Entropy ----")
print("BCE :", bce_result.numpy())

# Manual BCE
manual_bce = tf.reduce_mean(
    -(
        actual_binary * tf.math.log(prediction_binary)
        + (1 - actual_binary) * tf.math.log(1 - prediction_binary)
    )
)

print("Manual BCE :", manual_bce.numpy())


# ============================================================
# 3. CATEGORICAL CROSS-ENTROPY
# ============================================================

# Classes:
# 0 = Cat
# 1 = Dog
# 2 = Horse

# Actual = Dog
# One-hot encoded label = [0, 1, 0]

actual_categorical = tf.constant([[0.0, 1.0, 0.0]])

# Model prediction:
# Cat = 0.1
# Dog = 0.8
# Horse = 0.1

prediction_categorical = tf.constant([[0.1, 0.8, 0.1]])

cce = tf.keras.losses.CategoricalCrossentropy()

cce_result = cce(
    actual_categorical,
    prediction_categorical
)

print("\n---- Categorical Cross-Entropy ----")
print("CCE :", cce_result.numpy())

# Manual CCE
manual_cce = -tf.math.log(0.8)

print("Manual CCE :", manual_cce.numpy())


# ============================================================
# 4. SPARSE CATEGORICAL CROSS-ENTROPY
# ============================================================

# Instead of one-hot:
# Dog = class index 1

actual_sparse = tf.constant([1])

scce = tf.keras.losses.SparseCategoricalCrossentropy()

scce_result = scce(
    actual_sparse,
    prediction_categorical
)

print("\n---- Sparse Categorical Cross-Entropy ----")
print("Sparse CCE :", scce_result.numpy())


# ============================================================
# 5. SUMMARY
# ============================================================

print("\n========== SUMMARY ==========")

print("MSE                  :", mse_result.numpy())
print("MAE                  :", mae_result.numpy())
print("Huber                :", huber_result.numpy())
print("Binary CrossEntropy  :", bce_result.numpy())
print("Categorical CE       :", cce_result.numpy())
print("Sparse Categorical CE:", scce_result.numpy())

print("\n========== LABEL FORMAT ==========")

print("Categorical CE:")
print("One-hot label → [0, 1, 0]")

print("\nSparse Categorical CE:")
print("Integer label → 1")
```
